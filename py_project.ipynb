```python
import math


class Calculator:

    def __init__(self):
        # Dictionary containing the available operations
        self.operations = {
            "+": lambda a, b: a + b,
            "-": lambda a, b: a - b,
            "*": lambda a, b: a * b,
            "/": lambda a, b: a / b
        }

    def add_operation(self, symbol, function):
        # Add a new operation to the dictionary
        self.operations[symbol] = function

    def calculate(self, first_number, operation, second_number):

        # Check that both inputs are numbers
        if not isinstance(first_number, (int, float)):
            print("Error: the first value must be a number.")
            raise TypeError("First value must be a number.")

        if not isinstance(second_number, (int, float)):
            print("Error: the second value must be a number.")
            raise TypeError("Second value must be a number.")

        # Check that the operation exists
        if operation not in self.operations:
            print("Error: invalid operation.")
            raise ValueError("Invalid operation.")

        # Perform the calculation
        return self.operations[operation](first_number, second_number)


# Advanced mathematical operations

def exponentiation(a, b):
    return a ** b


def square_root(a, b):
    return math.sqrt(a)


def logarithm(a, b):
    return math.log(a)


# Create the calculator
calculator = Calculator()

# Add advanced operations
calculator.add_operation("**", exponentiation)
calculator.add_operation("sqrt", square_root)
calculator.add_operation("log", logarithm)


# Main program
while True:

    print("\nAvailable operations:")
    print("+  Addition")
    print("-  Subtraction")
    print("*  Multiplication")
    print("/  Division")
    print("** Exponentiation")
    print("sqrt Square root")
    print("log  Logarithm")
    print("q   Quit")

    operation = input("\nEnter an operation: ")

    # Exit the program
    if operation == "q":
        print("Goodbye!")
        break

    try:
        first_number = float(input("Enter the first number: "))

        # sqrt and log only need one mathematical input
        if operation in ["sqrt", "log"]:
            second_number = 0
        else:
            second_number = float(input("Enter the second number: "))

        result = calculator.calculate(
            first_number,
            operation,
            second_number
        )

        print("Result:", result)

    except Exception as error:
        print("Error:", error)
```
